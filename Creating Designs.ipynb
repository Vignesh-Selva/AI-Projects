## Imports
Import essential libraries for the image generation platform:
- gradio: For building the web interface
- openai: For accessing OpenAI's DALL-E API
- requests: For fetching image data via HTTP
- PIL (Python Imaging Library): For image processing
- io.BytesIO: For managing binary data streams

```python
import os
import requests
import gradio as gr
from PIL import Image
from io import BytesIO
from openai import OpenAI

print("All libraries imported successfully!")
```

## OpenAI API configuration

```python
# Initialize OpenAI client
client = OpenAI(
    api_key='your-api-key-here'
)

print("OpenAI client configured successfully!")
```

## Image Generation Function
This function generates images from text prompts using OpenAI's DALL-E API

```python
def generate_image(prompt):

    try:
        # Validate input
        if not prompt or prompt.strip() == "":
            return "Please enter a valid prompt description."

        # Call OpenAI's DALL-E API to generate image
        response = client.images.generate(model="dall-e-3", prompt=prompt, size="1024x1024", quality="standard", n=1)

        # Extract the image URL from the response
        image_url = response.data[0].url

        # Fetch the image data from the URL
        image_response = requests.get(image_url)

        # Check if the request was successful
        if image_response.status_code == 200:
            # Convert the binary data to an image
            image_data = BytesIO(image_response.content)
            generated_image = Image.open(image_data)
            return generated_image
        else:
            return f"Error fetching image: HTTP {image_response.status_code}"
    except Exception as e:
        return f"Error generating image: {str(e)}"

print("Image generation function created successfully!")
```

## Gradio Interface

Build the web interface using Gradio for user interaction

```python
# Define custom CSS for Netflix-themed styling
custom_css = """
    .gradio-container {
        font-family: 'Helvetica Neue', Arial, sans-serif;
        background: linear-gradient(135deg, #1a1a1a 0%, #2d0a0a 100%);
    }
    .title {
        color: #E50914;
        text-align: center;
        font-size: 2.5em;
        font-weight: bold;
        margin-bottom: 10px;
    }
    .description {
        color: #ffffff;
        text-align: center;
        font-size: 1.1em;
        margin-bottom: 30px;
    }
"""
```

```python
# Create the Gradio interface
demo = gr.Interface(
    fn=generate_image,
    # Input component: Textbox for entering prompts
    inputs=gr.Textbox(
        label="Enter Your Creative Prompt",
        placeholder="Example: A cinematic poster for a Netflix thriller series featuring a mysterious detective in a neon-lit city...",
        lines=5,
        max_lines=10,
    ),
    # Output component: Image display
    outputs=gr.Image(label="Generated Design", type="pil"),
    # Interface configuration
    title="🎬 Netflix Campaign Designer - AI Image Generator",
    description="""
    **Transform your creative vision into stunning visuals!**

    Enter a detailed text description and watch as AI generates custom images for Netflix marketing campaigns.
    Perfect for banners, posters, and promotional materials.

    **Tips for best results:**
    - Be specific and descriptive
    - Include style, mood, and visual elements
    - Mention "Netflix style" or "cinematic" for campaign-ready designs
    """,
    # Additional configuration
    theme=gr.themes.Soft(),
    examples=[
        ["A dramatic Netflix movie poster featuring a lone astronaut floating in space with Earth in the background, cinematic lighting, red and black color scheme"],
        ["A vibrant Netflix series banner showing a group of friends in a retro 80s arcade, neon lights, nostalgic atmosphere"],
        ["A mysterious Netflix thriller poster with a silhouette of a person standing in a foggy forest, dark and suspenseful mood"],
        ["An elegant Netflix documentary cover featuring wildlife photography of a majestic tiger in its natural habitat, professional National Geographic style"],
        ["A colorful Netflix animation poster with cartoon characters on an adventure, family-friendly, bright and cheerful"],
    ],
    # Enable caching for faster repeated generations
    cache_examples=False,
    # Interface appearance
    allow_flagging="never",
)

print("Gradio interface created successfully!")
```

## Launch the Application
Launch the Gradio interface to make it available for user interaction

```python
# Launch the interface
demo.launch(inline=True)

print("Application launched successfully!")
print("Access the interface through the URL displayed above")
```

## Enhanced Image Generation Function
Enhanced version with more customization options

```python
def generate_image_advanced(prompt, size, quality, style):

    try:
        if not prompt or prompt.strip() == "":
            return "Please enter a valid prompt description."

        response = client.images.generate(
            model="dall-e-3",
            prompt=prompt,
            size=size,
            quality=quality,
            style=style,
            n=1,
        )

        image_url = response.data[0].url
        image_response = requests.get(image_url)

        if image_response.status_code == 200:
            image_data = BytesIO(image_response.content)
            generated_image = Image.open(image_data)
            return generated_image
        else:
            return f"Error fetching image: HTTP {image_response.status_code}"

    except Exception as e:
        return f"Error generating image: {str(e)}"
```

## Create enhanced interface

```python
demo_advanced = gr.Interface(
    fn=generate_image_advanced,
    inputs=[
        gr.Textbox(
            label="Creative Prompt",
            placeholder="Describe your Netflix campaign visual...",
            lines=5,
        ),
        gr.Dropdown(
            choices=["1024x1024", "1024x1792", "1792x1024"],
            value="1024x1024",
            label="Image Size",
        ),
        gr.Radio(choices=["standard", "hd"], value="standard", label="Quality"),
        gr.Radio(choices=["vivid", "natural"], value="vivid", label="Style"),
    ],
    outputs=gr.Image(label="Generated Design", type="pil"),
    title="🎬 Netflix Campaign Designer - Advanced Mode",
    description="Create custom visuals with full control over size, quality, and style",
    theme=gr.themes.Soft(),
)

print("Enhanced interface created!")
```
